1.- Dos equipos juegan una serie de juegos (4 de 7, el que gane 4 juegos gana). Cada equipo tiene un 50% de probabilidad de ganar cada juego (no empates). ¿Cuál es la probabilidad de que la serie se vaya a 7 juegos? (5/16)

```python
import numpy as np
```

```python
def simulacion_montecarlo(n_simulaciones):
    series_a_siete = 0
    for i in range(n_simulaciones):
        victorias_a = 0
        victorias_b = 0
        for i in range(1, 8):
            quien_gano = np.random.randint(0, 2)
            if quien_gano == 0:
                victorias_a += 1
            else:
                victorias_b += 1

            if (victorias_a == 4) or (victorias_b == 4) :
              if i==7:
                series_a_siete += 1
              break

    probabilidad = series_a_siete / n_simulaciones
    print(f'{probabilidad:.4f}')

```

```python
simulacion_montecarlo(n_simulaciones=1000000)
```

2.- Una en mil personas tienen una enfermedad H, y la prueba de la enfermedad sale correcta 98% de las veces cuando la persona está enferma. Por otro lado, si la persona está enferma la prueba tiene un error del 1%. Si alguien se hace una prueba y sale positiva, ¿cuál es la probabilidad de que tengan la enfermedad? (98/1097)

```python
def proba_enf_dado_postivo(n_personas):
  enf_pos= 0
  tot_pos= 0
  for i in range(n_personas):
    esta_enfermo = np.random.random() < 0.001
    resultado_positivo = False
    rand_prueba = np.random.random()
    if esta_enfermo:
      if rand_prueba < 0.98:
          resultado_positivo = True
    else:
        if rand_prueba < 0.01:
            resultado_positivo = True
    if resultado_positivo:
        tot_pos += 1
        if esta_enfermo:
            enf_pos += 1

  return enf_pos / tot_pos

```

```python
proba_enf_dado_postivo(n_personas=10000000)
```
