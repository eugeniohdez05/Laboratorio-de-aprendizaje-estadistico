```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
import numpy as np
import statsmodels.api as sm
import random
import warnings
warnings.filterwarnings('ignore')
from sklearn.metrics import mean_squared_error, r2_score
```

```python
df = pd.read_excel("Motor Trend Car Road Tests.xlsx")
df.head()
```

```python
lr= LinearRegression()
X= df[['hp', 'qsec']]
y= df['mpg']
```

```python
lr.fit(X,y)
```

```python
B0= lr.intercept_
B1= lr.coef_[0]
B2= lr.coef_[1]
print(f'B0 = {B0}')
print(f'B1 = {B1}')
print(f'B2 = {B2}')
```

```python
#Usando ols
X = sm.add_constant(X)
model=sm.OLS(y, X) # -> Modela minimos cuadrados ordinarios con todas las variables
results=model.fit() #-> Optimiza
print(results.summary())
```

```python
B0= results.bse[0]
B1= results.bse[1]
B2= results.bse[2]
print(f'B0 = {B0}')
print(f'B1 = {B1}')
print(f'B2 = {B2}')
```

```python
#Intervalo de confianza
results.conf_int(alpha=0.05, cols=None)
```

```python
B0=[]
B1=[]
B2=[]
for i in range (1001):
  df_boot= df.sample(df.shape[0], replace=True)
  X= df_boot[['hp', 'qsec']]
  y= df_boot['mpg']
  X= sm.add_constant(X)
  model=sm.OLS(y, X)
  results=model.fit()
  #guardar cada regresión
  B0.append(results.bse[0])
  B1.append(results.bse[1])
  B2.append(results.bse[2])
```

# Aggregating

```python
modelos = []
R2 = []
vars= []
models_test = []
```

```python

for i in range (1001):
  available_features = ['cyl', 'disp', 'hp', 'drat', 'wt', 'qsec', 'vs', 'am', 'gear', 'carb']
  random_cols = random.sample(available_features, 3)
  X = df[random_cols]
  y = df['mpg']
  X= sm.add_constant(X)
  X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.5, random_state=42)
  model=sm.OLS(y_train, X_train)
  results=model.fit()
  modelos += [results]
  vars.append(random_cols)
  results_test= results.predict(X_test)
  models_test.append(results_test)



```

```python
y_test= np.array(y_test).reshape(-1,1)
```

```python
mean_models_test= np.mean(models_test, axis=0)
mean_models_test= np.array(mean_models_test).reshape(-1,1)
mean_models_test
R2= r2_score(y_test, mean_models_test)
R2
```
