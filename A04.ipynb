```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.gaussian_process import GaussianProcessRegressor
from sklearn.gaussian_process.kernels import RBF
```

```python
y= [2847,-4017, 1255, 1773,563]
x= np.array([10,2,3,7,12]).reshape(-1, 1)
```

```python

gp= GaussianProcessRegressor (n_restarts_optimizer=10, random_state=42)
gp.fit(x,y)
```

```python
x_sim= np.linspace(2, 15, 10001).reshape(-1,1)
y_pred, sigma= gp.predict(x_sim, return_std=True)
x_sim
```

```python
print(f'{y_pred}, {sigma}')
```

```python
#graficar y_pred
plt.figure(figsize=(10, 6))
plt.plot(x_sim, y_pred, label='Predicción')
plt.xlabel('x')
plt.ylabel('y')
plt.title('Predicción de y')
plt.legend
plt.show()
```

```python
y_upper= y_pred + 2*sigma
y_lower= y_pred - 2*sigma
```

```python
#nueva grafica con y_upper y y_lower
plt.figure(figsize=(10, 6))
plt.fill_between(x_sim.ravel(), y_upper.ravel(), y_lower.ravel(), alpha=0.3 , label='Intervalo de confianza', color='darkred')
plt.plot(x_sim, y_pred, label='Predicción')
plt.ylim([0,3000])
plt.xlabel('x')
plt.ylabel('y')
plt.title('Predicción de y')
plt.legend()
plt.show()
```

```python
idx_max_incertidumbre= np.argmax(y_pred)
x_new_sample= x_sim[idx_max_incertidumbre].reshape(1,1)
y_new_sample= y_pred[idx_max_incertidumbre]
print(f'{x_new_sample,y_new_sample}')
x = np.vstack([x, x_new_sample])
y = np.append(y, 2847.31363)
```
