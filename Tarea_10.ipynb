### Conceptos


*   **GINI**:  Es una medida de la varianza total a través de las *k* clases. Se define matemáticamente como: $G = \sum_{k=1}^{K} \hat{p}_{mk}(1 - \hat{p}_{mk})$, donde $\hat p_{mk}$ es la proporción de observaciones en el nodo *m* que pertenecen a la clase *k*.
*   **Entropía**: Al igual que el método gini, la entropía también mide la pureza del nodo (pureza refiriendose a si el nodo tiene una sola clase o más); su fórmula es:
$D = -\sum_{k=1}^{K} \hat{p}_{mk} \log \hat{p}_{mk}$ la entropía toma valores cercanos a cero si las proporciones $\hat{p}_{mk}$ están cerca de cero o de uno
*   **Log Loss**: También es conocido como entropía cruzada, es usado principalmente en clasificación, ya que mide la disimilitud entre las probabilidades predichas y las etiquetas reales. Un log loss menor indica mejores predicciones y un modelo más preciso, siendo cero en un modelo perfecto.
Su fórmula es:
 $$J = -\frac{1}{N} \sum_{i=1}^{N} [y_i \log(p_i) + (1 - y_i) \log(1 - p_i)]$$
*   **¿Cuál es la diferencia entre entropía y log loss?**: Log loss suele ser aplicado para casos de clasificación binaria, mientras que entropía se suele usar para clasificación binaria y también multiclase, osea es mucho más general.

```python
import pandas as pd
import numpy as np
import sklearn as sk
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#Partición GINI

def gini(data_col):
  n = len(data_col)
  if n == 0: return 0
  v, counts = np.unique(data_col, return_counts=True)
  probabilidades = counts / n
  gini = 1 - np.sum(probabilidades**2)
  return gini
```

```python
#Partición Entropía

def Entropía(data_col):
  n = len(data_col)
  if n == 0: return 0
  v, counts = np.unique(data_col, return_counts=True)
  probabilidades = counts / n
  entropía = -np.sum(probabilidades*np.log2(probabilidades))
  return entropía
```

```python
#Partición Log Loss

def log_loss(data_col):
  n = len(data_col)
  if n == 0: return 0
  v, counts = np.unique(data_col, return_counts= True)
  probabilidades = counts / n
  log_loss = -np.sum(probabilidades*np.log2(probabilidades))
  return log_loss
```
