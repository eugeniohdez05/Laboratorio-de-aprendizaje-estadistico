### Regresión Lineal Simple


```python
lr= LinearRegression()
X= df2.drop(columns=['sales'])
y= df2['sales']
lr.fit(X, y)
B0 = lr.intercept_
B1= lr.coef_[0]
B2= lr.coef_[1]
B3= lr.coef_[2]
print(f'B0: {B0:.4f}, newspaper: {B1:.4f}, TV: {B2:.4f}, radio: {B3:.4f}')
R2= lr.score(X, y)
print(f'R2: {R2:.5f}')
```

```python
#Usando pipeline
pipeline = Pipeline([
    ('regresion', LinearRegression())
])
pipeline.fit(X, y)
B0 = pipeline.named_steps['regresion'].intercept_
B1= pipeline.named_steps['regresion'].coef_[0]
B2= pipeline.named_steps['regresion'].coef_[1]
B3= pipeline.named_steps['regresion'].coef_[2]
print(f'B0: {B0:.4f}, newspaper: {B1:.4f}, TV: {B2:.4f}, radio: {B3:.4f}')
R2= pipeline.score(X, y)
print(f'R2: {R2:.5f}')
```

# Árbol de decisión

```python
arbol = DecisionTreeRegressor()
arbol.fit(X, y)
pipeline_arbol = Pipeline([
    ('arbol', DecisionTreeRegressor())
])
pipeline_arbol.fit(X, y)
R2_pipeline_arbol = pipeline_arbol.score(X, y)
print(f'R2 del árbol de decisión: {R2_pipeline_arbol:.5f}')
```

```python
#Usando train test split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.5, random_state=42)
pipeline_arbol.fit(X_train, y_train)
R2_pipeline_arbol_test = pipeline_arbol.score(X_test, y_test)
print(f'R2 del árbol de decisión en el conjunto de entrenamiento: {R2_pipeline_arbol:.5f}')
print(f'R2 del árbol de decisión en el conjunto de prueba: {R2_pipeline_arbol_test:.5f}')
```

```python
pipeline.fit(X_train, y_train)
R2_pipeline = pipeline.score(X_test, y_test)
print(f'R2 del modelo de regresión lineal en el conjunto de prueba: {R2_pipeline:.5f}')
```
